```python
import pandas as pd
import sqlalchemy

engine = sqlalchemy.create_engine("sqlite+pysqlite:///data/orders_and_users.db")
inspector = sqlalchemy.inspect(engine)
print(inspector.get_table_names())
inspector.get_columns("orders")
```

```python
query = sqlalchemy.text("""
SELECT customers.Location, orders.Category, COUNT(*) AS orders_count
FROM orders
INNER JOIN customers
ON orders.CustomerID = customers.CustomerID
GROUP BY customers.Location, orders.Category
ORDER BY customers.Location, orders_count DESC
""")
orders_by_location_category = pd.read_sql(query, con=engine)
orders_by_location_category
```

```python
query = sqlalchemy.text("""
SELECT customers.Location, COUNT(*) AS keyboard_orders
FROM orders
INNER JOIN customers
ON orders.CustomerID = customers.CustomerID
WHERE orders.Description = 'Gaming Keyboard'
GROUP BY customers.Location
ORDER BY keyboard_orders DESC
""")
keyboards_by_location = pd.read_sql(query, con=engine)
keyboards_by_location
```

```python
query = sqlalchemy.text("""
SELECT orders.Category, COUNT(*) AS orders_count
FROM orders
INNER JOIN customers
ON orders.CustomerID = customers.CustomerID
WHERE customers.Annual_Income > 2000
GROUP BY orders.Category
ORDER BY orders_count DESC
""")
category_income_2000 = pd.read_sql(query, con=engine)
category_income_2000
```
