```python
import pandas as pd

apartments_df = pd.read_csv("data/appartments_data_set.csv")
display(apartments_df.head(10))
```

```python
apartments_df.info()
```

```python
apartments_df.drop(columns=["Unnamed: 0"], inplace=True)
apartments_df.fillna({"ceiling_height": apartments_df["ceiling_height"].median()}, inplace=True)
apartments_df.info()
```

```python
apartments_df.dropna(subset=["floors_total"], inplace=True)
apartments_df.drop(columns=["is_apartment"], inplace=True)
apartments_df.info()
```

```python
ratio = (apartments_df["living_area"] / apartments_df["total_area"]).median()
apartments_df["living_area"] = apartments_df["living_area"].fillna(apartments_df["total_area"] * ratio)
print("ratio = ", ratio)
apartments_df.info()
```

```python
apartments_df.fillna({
    "kitchen_area": apartments_df["kitchen_area"].median(),
    "balcony": 0
}, inplace=True)
apartments_df.drop(columns=["days_exposition"], inplace=True)
apartments_df.info()
display(apartments_df.head())
```

```python
apartments_df["first_day_exposition"] = pd.to_datetime(apartments_df["first_day_exposition"])
apartments_df = apartments_df.astype({
    "last_price": "int64",
    "floors_total": "int64",
    "balcony": "int64",
    "locality_name": "category",
})
apartments_df.info()
```

```python
apartments_df.duplicated(keep="first").sum()
```

```python
apartments_df["price_per_sqm"] = apartments_df["last_price"] / apartments_df["total_area"]
display(apartments_df.head())
```

```python
import data_profiling

profiler = data_profiling.ProfileReport(apartments_df)
profiler.to_file("data/apartments_report.html")
```
