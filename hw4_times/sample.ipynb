```python
import pandas as pd

apartments_df = pd.read_csv("data/processed_apartments_data_set.csv")
display(apartments_df.head(10))
apartments_df.info()
```

```python
import datetime
apartments_df["first_day_exposition"] = pd.to_datetime(apartments_df["first_day_exposition"])
apartments_df.info()
```

```python
apartments_df.sort_values(by="first_day_exposition", inplace = True)
apartments_df.set_index("first_day_exposition", drop=True,inplace = True)
apartments_df.head(10)
```

```python
year_2023_df = apartments_df.loc["2023"]
year_2023_df[year_2023_df["rooms"] == 2].pivot_table(
    values="rooms",
    index="locality_name",
    aggfunc="count"
).sort_values(by="rooms", ascending=False)
```

```python
kyiv_3rooms = apartments_df[
    (apartments_df["rooms"] == 3) & (apartments_df["locality_name"] == "Kyiv")
]
kyiv_3rooms = kyiv_3rooms.resample("YE")["total_area"].count()
kyiv_3rooms
```

```python
kyiv_df = apartments_df[apartments_df["locality_name"] == "Kyiv"]
kyiv_rooms_dynamics = kyiv_df.groupby("rooms").resample("YE")["total_area"].count()
kyiv_rooms_dynamics
```

```python
apartments_df.to_csv("data/processed_apartments_data_set.csv", index=False)
```
