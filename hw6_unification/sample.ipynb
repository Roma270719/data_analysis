```python
import pandas as pd

customers_df = pd.read_csv("data/customers.csv")
orders_part_1_df = pd.read_csv("data/orders_part_1.csv")
orders_part_2_df = pd.read_csv("data/orders_part_2.csv")
display(customers_df.head(10))
display(orders_part_1_df.head(10))
display(orders_part_2_df.head(10))
```

```python
orders = pd.concat([orders_part_1_df, orders_part_2_df], axis=0, ignore_index=True)
orders
```

```python
orders.duplicated().sum()
```

```python
full_df = orders.merge(customers_df, how="left", on="CustomerID")
full_df
```

```python
task1 = full_df.groupby(["Location", "Category"]).size().reset_index(name="OrderCount")
task1
```

```python
keyboards = full_df[full_df["Description"] == "Gaming Keyboard"]
task2 = keyboards.groupby("Location").size().sort_values(ascending=False)
task2
```

```python
rich_customers = full_df[full_df["Annual_Income"] > 2000]
task3 = rich_customers.groupby("Category").size().sort_values(ascending=False)
task3
```
