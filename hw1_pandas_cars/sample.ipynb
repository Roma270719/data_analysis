```python
import pandas as pd

cars_df = pd.read_csv("data/cars_data_set.csv")
display(cars_df.head())
```

```python
cars_df.columns = cars_df.columns.str.lower().str.replace(' ', '_')
display(cars_df.head())
```

```python
cars_df.info()
```

```python
cars_df["market_category"].unique()
```

```python
cars_df["market_category"] = cars_df["market_category"].fillna("unknown")
cars_df.info()
```

```python
cars_df.fillna({"engine_hp": cars_df["engine_hp"].median()}, inplace=True)
cars_df.info()
```

```python
cars_df.loc[(cars_df["engine_cylinders"].isna()) & (cars_df["engine_fuel_type"]=="electric"), ["engine_cylinders"]] = 0
cars_df.fillna({"engine_cylinders": cars_df["engine_cylinders"].median()}, inplace=True)
cars_df.dropna(subset=["number_of_doors"], inplace=True)
cars_df.info()
```

```python
cars_df.duplicated(keep= "first").sum()
```

```python
cars_df.drop_duplicates(keep= "first", inplace=True)
cars_df.info()
cars_df.duplicated(keep= "first").sum()
```

```python
cars_df = cars_df.astype({
    "make": "category",
    "model": "category",
    "year": "int64",
    "engine_fuel_type": "category",
    "engine_hp": "int64",
    "engine_cylinders": "int64",
    "transmission_type": "category",
    "driven_wheels": "category",
    "number_of_doors": "int64",
    "market_category": "category",
    "vehicle_size": "category",
    "vehicle_style": "category",
    "highway_mpg": "int64",
    "city_mpg": "int64",
    "popularity": "int64",
    "msrp": "int64",
})
cars_df.info()
```

```python
cars_df.to_csv("data/processed_cars_data_set.csv", sep=",", index=False)
```

```python
cars_df["price_per_hp"] = cars_df["msrp"] / cars_df["engine_hp"]
display(cars_df.head())
```

```python
bmw_cars = cars_df[cars_df["make"] == "BMW"]
top5_bmw = bmw_cars.nlargest(5, "msrp")
display(top5_bmw)
```
