```python
import pandas as pd

distributor_df = pd.read_csv("data/distributor-info.csv")
reclama_df = pd.read_csv("data/reclamni_zosobi-info.csv")
display(distributor_df.head(10))
distributor_df.info()
display(reclama_df.head(10))
reclama_df.info()
```

```python
import datetime

reclama_df = reclama_df.dropna(subset=["planesValue", "horizontalSize", "verticalSize"])
reclama_df = reclama_df.dropna(subset=["planesValue"]).reset_index(drop=True)
reclama_df = reclama_df.astype({
    "status": "category",
    "type": "category",
    "district": "category",
    "planesValue": "int64",
})

reclama_df["issuedDate"] = pd.to_datetime(reclama_df["issuedDate"])
reclama_df["validFrom"] = pd.to_datetime(reclama_df["validFrom"])
reclama_df["validThrough"] = pd.to_datetime(reclama_df["validThrough"])

reclama_df.info()
```

```python
import data_profiling

profiler = data_profiling.ProfileReport(reclama_df)
profiler.to_file("data/reclama_report.html")
```

```python
outlier_cols = ['horizontalSize', 'verticalSize', 'area', 'planesValue']

for col in outlier_cols:
    q1 = reclama_df.groupby('type', observed=True)[col].transform('quantile', 0.25)
    q3 = reclama_df.groupby('type', observed=True)[col].transform('quantile', 0.75)
    iqr = q3 - q1

    lower = q1 - 3 * iqr
    upper = q3 + 3 * iqr

    reclama_df[f'{col}_is_outlier'] = ((reclama_df[col] < lower) | (reclama_df[col] > upper)) & (iqr > 0)

outlier_flag_cols = [f'{col}_is_outlier' for col in outlier_cols]
reclama_df['is_outlier'] = reclama_df[outlier_flag_cols].any(axis=1)

print(reclama_df['is_outlier'].sum(), 'из', len(reclama_df))
```

```python
reclama_df.groupby('type', observed=True)['is_outlier'].agg(['sum', 'mean', 'size']).sort_values('mean', ascending=False).head(15)
reclama_df[reclama_df['is_outlier']].nlargest(15, 'area')[['type', 'area', 'horizontalSize', 'verticalSize', 'planesValue']]
reclama_df[reclama_df['is_outlier']].nlargest(15, 'horizontalSize')[['type', 'area', 'horizontalSize', 'verticalSize', 'planesValue']]
```

```python
reclama_df['hw_ratio'] = reclama_df['verticalSize'] / reclama_df['horizontalSize']

suspicious = reclama_df[reclama_df['is_outlier'] & (
    (reclama_df['hw_ratio'] > 5) | (reclama_df['hw_ratio'] < 0.2)
)]
print(len(suspicious))
suspicious[['type', 'horizontalSize', 'verticalSize', 'area', 'hw_ratio']].sort_values('hw_ratio', ascending=False)
```

```python
suspicious.sort_values('hw_ratio', ascending=True).head(10)
```

```python
standardized_types = [
    'Кронштейн на фасаді будинку (будівлі), споруді',
    'Тумба, що стоїть окремо',
    "Афішна (інформаційна) тумба",
    'Прапор, прапорець, який використовується як рекламоносій, що стоїть окремо',
]

suspicious_final = suspicious[suspicious['type'].isin(standardized_types)]
print(len(suspicious_final))
suspicious_final[['type','horizontalSize','verticalSize','area','hw_ratio']]
```

```python
import numpy as np

today = pd.Timestamp.now().normalize()
reclama_df['status_actual'] = np.where(
    reclama_df['validThrough'] < today,
    'Недійсний (термін минув)',
    'Дійсний'
)
reclama_df = reclama_df.astype({"status_actual": "category"})
comparison = pd.crosstab(reclama_df['status'], reclama_df['status_actual'])
print(comparison)
reclama_df.info()
reclama_df.duplicated(keep="first").sum()
reclama_df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
active_df = reclama_df[reclama_df["status_actual"] == "Дійсний"]
by_planes = active_df.groupby("type", observed=True)["planesValue"].sum().sort_values(ascending=False).head(10)
print(by_planes)
fig = px.bar(
    by_planes.sort_values(ascending=True),
    orientation='h',
    title='Топ-10 типів зовнішньої реклами по місту (к-сть поверхонь)',
    labels={'value': 'Кількість поверхонь', 'type': ''}
)
fig.show()
```

```python
pivot_type_district = active_df.pivot_table(
    index="district",
    columns="type",
    values="planesValue",
    aggfunc="sum",
    fill_value=0,
    observed=True
)
top_type_per_district = pivot_type_district.idxmax(axis=1)
top_type_value = pivot_type_district.max(axis=1)

result = pd.DataFrame({
    'top_type': top_type_per_district,
    'planesValue': top_type_value
})
print(result)

fig = px.bar(
    result.reset_index(),
    x='district',
    y='planesValue',
    color='top_type',
    title='Найпопулярніший тип реклами в кожному районі',
    labels={'planesValue': 'Кількість поверхонь', 'district': 'Район', 'top_type': 'Тип реклами'}
)
fig.show()
```

```python
merged_df = active_df.merge(
    distributor_df,
    left_on='distributorid',
    right_on='id',
    how='left',
    suffixes=('', '_distributor')
)
print(len(active_df), len(merged_df))
merged_df
```

```python
top_companies_planes = (
    merged_df.groupby('distributor', observed=True)['planesValue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
print(top_companies_planes)

import re

def short_name(name):
    match = re.search(r'«(.+?)»', name)
    return match.group(1) if match else name[:40]

top_companies_short = top_companies_planes.copy()
top_companies_short.index = top_companies_short.index.map(short_name)

fig = px.bar(
    top_companies_short.sort_values(ascending=True),
    orientation='h',
    title='Топ-10 компаній за кількістю рекламних поверхонь (по місту)',
    labels={'value': 'Кількість поверхонь', 'index': 'Компанія'}
)
fig.update_layout(showlegend=False)
fig.show()
```

```python
top_companies_area = (
    merged_df.groupby('distributor', observed=True)['area']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
print(top_companies_area)

top_companies_area_short = top_companies_area.copy()
top_companies_area_short.index = top_companies_area_short.index.map(short_name)

fig = px.bar(
    top_companies_area_short.sort_values(ascending=True),
    orientation='h',
    title='Топ-10 компаній за площею зовнішньої реклами (по місту)',
    labels={'value': 'Площа, м²', 'index': 'Компанія'}
)
fig.update_layout(showlegend=False)
fig.show()
```

```python
pivot_company_district = merged_df.pivot_table(
    index='district',
    columns='distributor',
    values='area',
    aggfunc='sum',
    fill_value=0,
    observed=True
)

top_company_per_district = pivot_company_district.idxmax(axis=1)
top_company_value = pivot_company_district.max(axis=1)

result_company = pd.DataFrame({
    'top_company': top_company_per_district.map(short_name),
    'area': top_company_value
})
print(result_company)

result_company_sorted = result_company.sort_values('area', ascending=True)

fig = px.bar(
    result_company_sorted.reset_index(),
    x='area',
    y='district',
    color='top_company',
    orientation='h',
    title='Компанія-лідер за площею реклами в кожному районі',
    labels={'area': 'Площа, м²', 'district': 'Район', 'top_company': 'Компанія'}
)
fig.show()
```

```python
merged_df["days_left"] = (merged_df["validThrough"] - today).dt.days

result = merged_df.sort_values("days_left")[
    ["distributor", "days_left"]
].reset_index(drop=True)

top20 = result.head(20)
print(top20)
```

```python
top5 = merged_df["distributor"].value_counts().head(5).index
data = merged_df[merged_df["distributor"].isin(top5)].assign(
    distributor_short=lambda d: d["distributor"].map(short_name)
)

ct = pd.crosstab(data["distributor_short"], data["type"].astype(str))

top_type_name = ct.idxmax(axis=1)   # <-- вот эта строка недостающая
top_type_value = ct.max(axis=1)

chart_df = pd.DataFrame({
    "топ-тип": top_type_value,
    "інше": ct.sum(axis=1) - top_type_value,
})
print(pd.DataFrame({"top_type": top_type_name, "count": top_type_value}))  # печатаем название

fig = px.bar(chart_df, title="Найпопулярніший тип реклами у топ-5 компаній")
fig.show()
```

```python
darnytskyi = merged_df[merged_df["district"] == "Дарницький"]
top3_names = darnytskyi["distributor"].value_counts().head(3).index.tolist()

leaders = darnytskyi[darnytskyi["distributor"].isin(top3_names)].copy()
leaders["distributor_short"] = leaders["distributor"].str.extract(r"«(.+?)»")

type_counts = leaders.groupby(["distributor_short", "type"], observed=True).size().reset_index(name="count")
type_counts = type_counts[type_counts["count"] > 0]

for company, df_company in type_counts.groupby("distributor_short"):
    fig = px.pie(
        df_company, names="type", values="count",
        title=company,
    )
    fig.update_traces(textinfo="percent+value", textposition="inside")
    fig.update_layout(
        height=700,
        width=800,
        legend=dict(orientation="h", yanchor="bottom", y=-0.4, x=0),
        margin=dict(t=80, b=200),
    )
    fig.show()
```

```python
merged_all = reclama_df.merge(distributor_df, left_on="distributorid", right_on="id", how="left")
darnytskyi_all = merged_all[merged_all["district"] == "Дарницький"].copy()
darnytskyi_all["year"] = darnytskyi_all["issuedDate"].dt.year

by_year = darnytskyi_all.groupby("year")["area"].mean().round(2).reset_index()

fig = px.line(
    by_year, x="year", y="area", markers=True,
    line_shape="spline",
    title="Середня площа однієї реклами за роком укладання договору (Дарницький)",
    labels={"year": "Рік", "area": "Середня площа, м²"}
)
fig.show()
```

```python
bigboard = merged_all[
    (merged_all["distributor"].str.contains("БІГБОРД", na=False)) &
    (merged_all["district"] == "Дарницький")
].copy()
bigboard["year"] = bigboard["issuedDate"].dt.year

# объём по годам
by_year = bigboard.groupby("year")["area"].sum().reset_index()
fig1 = px.bar(by_year, x="year", y="area", title="БІГБОРД: сумарна площа реклами за роком",
              labels={"year": "Рік", "area": "Площа, м²"})
by_year_type_area = bigboard.groupby(["year", "type"], observed=True)["area"].sum().reset_index()
by_year_type_area["year"] = by_year_type_area["year"].astype(str)

fig = px.bar(
    by_year_type_area, x="year", y="area", color="type",
    barmode="group",
    title="БІГБОРД: обсяг та типи реклами за роком (Дарницький)",
    labels={"year": "Рік", "area": "Площа, м²", "type": "Тип"}
)
fig.update_xaxes(type="category")
fig.show()
```
